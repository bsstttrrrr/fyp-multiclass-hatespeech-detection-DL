```python
import pipeline
import pandas as pd
import numpy as np
df=pd.read_csv('processed_tweets.csv')
X_train, X_val, X_test, y_train, y_val, y_test = pipeline.get_train_test_data(df)
```

```python
tuner, tokenizer = pipeline.run_hyperparameter_search(X_train, y_train, X_val, y_val)
best_hps, best_model = pipeline.get_best_hyperparams_and_model(tuner)
```

```python
X_full = X_train + X_val
y_full = np.concatenate([y_train, y_val])
history = pipeline.train_model( best_model, X_train, y_train, X_val, y_val, tokenizer, epochs=30, batch_size=16)
```

```python
pipeline.plot_training_history(history, model_name="Fine-tuned DistilBERT-CNN-BiLSTM")
```

```python
pipeline.evaluate_model(best_model, X_test, y_test, tokenizer, model_name="Fine-tuned DistilBERT-CNN-BiLSTM")
```

```python
auc_score = pipeline.plot_multiclass_roc(best_model, X_val, y_val, tokenizer, model_name="Fine-tuned DistilBERT-CNN-BiLSTM")
```
